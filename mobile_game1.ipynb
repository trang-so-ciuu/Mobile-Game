```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Gender'] = df['Gender'].fillna('Unknown')
df['Country'] = df['Country'].fillna('Unknown')
df['Device'] = df['Device'].fillna('Unknown')
df['GameGenre'] = df['GameGenre'].fillna('Unknown')
```

```python
df['LastPurchaseDate'] = pd.to_datetime(
    df['LastPurchaseDate'],
    errors='coerce'
)
type(df['LastPurchaseDate'].dtype)
```

```python
df.groupby('SpendingSegment')['InAppPurchaseAmount'].agg(
    ['count', 'min', 'max', 'mean', 'sum']
)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

connection_string = (
    r"DRIVER={ODBC Driver 17 for SQL Server};"
    r"SERVER=localhost\SQLEXPRESS;"
    "DATABASE=mobile_game;"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    "mssql+pyodbc:///?odbc_connect=" + quote_plus(connection_string)
)

df.to_sql("mobile_game_data",engine, if_exists="replace", index=False)
```
